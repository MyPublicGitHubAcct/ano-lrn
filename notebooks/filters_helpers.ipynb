# Filter Helpers (biquad internals)

This module provides the low-level coefficient builder used by every filter in `src/python/filters/`. Direct calls to these helpers are not normally needed — use the public filter functions in `filters/eq.py`, `filters/shelving.py`, etc.

**Key internal function:**

| Helper | Purpose |
| --- | --- |
| `_biquad_coeffs(b, a, signal)` | Apply a pre-computed `(b0, b1, b2, a1, a2)` biquad to a signal via `scipy.signal.lfilter`. All public filters delegate here after computing their specific coefficients. |
